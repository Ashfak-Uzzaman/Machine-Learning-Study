```python
import numpy as np
```

There is 20 training example and each has 5 features

**Note:** Here randomly values are choosen. It is only for implementation

```python
number_of_features = 5
number_of_examples = 20

first_column = np.ones((number_of_examples, 1))

X = np.random.randint(1, 101, size=(number_of_examples, number_of_features))
print(X)

X = np.concatenate((first_column, X), axis=1)
print(X.shape)
Y = np.random.randint(1, 101, size=number_of_examples)
```

```python
X_T = np.transpose(X)

W = X_T @ X
W = np.linalg.inv(W)    
W = W @ X_T

W = W @ Y

print(W)
```

```python
prediction = X @ W

prediction
```
