##### Sigmoid Activation Function

````python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
````

<img src="images/forward_prop.png" width="800" height="350">

````python
# dense layer:  A set of weights and bias
def dense(a_in,W,b):
    units=W.shape[1] # units are the nurons of a layer. Here  W[number_of_feature_a_in][number_of_units]

    a_out=np.zeros(units)

    for j in range(units):
        w=W[:,j]
        z=np.dot(a_in,w)+b[j] # a will dor with each units
        a_out[j] = sigmoid(z)

    return a_out
````

#### dense function using numpy **'matmul'** (matrix-multiplication function)

````python
def dense_matmul(X,W,b):
    z = np.matmul(X,W) + b
    a_out = sigmoid(z)
    return a_out
````

````python
W_tmp = [np.array([[-8.93,  0.29, 12.9 ], 
                   [-0.1,  -7.32, 10.81]]),  # 3 units, 2 weights for 2 features # layer-1

         np.array([[-31.18], 
                   [-27.59], 
                   [-32.56]])]  # 1 units, 3 weights for 3 features # layer-2 (output layer)


b_tmp = [np.array([-9.82, -9.28,  0.96]),  # 3 unit # layer-1
         np.array([15.41])]               # 1 unit # layer-2 (output layer)
````

````python
def sequential(x, W, b):
    layers=len(W)
    a = x
    for layer in range(layers):
        a = dense(a, W[layer], b[layer])
    return a
````

#### Predictions

````python
def my_predict(X, W, b):
    m = X.shape[0]
    p = np.zeros((m,1))
    for i in range(m):
        p[i,0] = 0 if sequential(X[i], W, b)<0.5 else 1
    return(p)
````

````python
prediction=my_predict(X_standardized,W_tmp,b_tmp)

print(prediction)

for i in range(len(Y)):
    print(Y[i]==prediction[i])
````
