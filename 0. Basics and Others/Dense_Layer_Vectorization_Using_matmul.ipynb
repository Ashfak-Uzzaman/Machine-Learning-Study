#### **Dense**

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

```python
def dense(X,W,b):
    z = np.matmul(X,W) + b
    a_out = sigmoid(z)
    return a_out
```

.

#### Calculating final output of the network

```python
def sequential(x, W, b):
    layers=len(W)
    a = x
    for layer in range(layers):
        a = dense(a, W[layer], b[layer])
    return a
```

#### Predictions

```python
def my_predict(X, W, b):
    m = X.shape[0]
    p = np.zeros((m,1))
    for i in range(m):
        p[i,0] = 0 if sequential(X[i], W, b)<0.5 else 1
    return(p)
```
