#  29. Write Python code to fill missing values with the mean of the column.

````python
df=pd.read_csv(f"data/NAN-HeightWeightBMI.csv")
df.head()
````

````python
df.columns
````

````python
# Fill missing values in numeric columns with the column mean
df['Height(Inches)'] = df['Height(Inches)'].fillna(df['Height(Inches)'].mean())
df['Weight(Pounds)'] = df['Weight(Pounds)'].fillna(df['Weight(Pounds)'].mean())
df['BMI'] = df['BMI'].fillna(df['BMI'].mean())
df.head()
````

# One-Hot Encoding

````python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from xgboost import XGBClassifier
````

````python
# Load the dataset using pandas
df = pd.read_csv("data\heart.csv")
df.head()
````

### One-hot encoding using Pandas

````python
# These columns will replace 
cat_variables = ['Sex',
'ChestPainType',
'RestingECG',
'ExerciseAngina',
'ST_Slope'
]
````

````python
# This will replace the columns with the one-hot encoded ones and keep the columns outside 'columns' argument as it is.
df = pd.get_dummies(data = df,
                         prefix = cat_variables,
                         columns = cat_variables)
````

````python
var = [x for x in df.columns if x not in 'HeartDisease'] ## Removing our target variable
````

````python
print(len(var))
````

### Splitting the Dataset

````python
RANDOM_STATE = 42  # Set your random state for reproducibility
X_train, X_test, y_train, y_test = train_test_split(df[var], df['HeartDisease'], train_size=0.8, random_state=RANDOM_STATE)
````

````python
print(f'train samples: {len(X_train)}\ntest samples: {len(X_test)}')
print(f'target proportion: {sum(y_train)/len(y_train):.4f}')
````

##### Display CSV

````python
import csv
from tabulate import tabulate

def display_csv_tabulate(filepath):
    with open(filepath, 'r', newline='') as csvfile:
        reader = csv.reader(csvfile)
        header = next(reader)
        rows = list(reader)
   
        print(tabulate(rows, headers=header, tablefmt="grid")) 
````

````python
display_csv_tabulate('data/output_pandas_numpy.csv')
````
