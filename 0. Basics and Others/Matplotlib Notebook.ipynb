```python
x_values = [1, 2, 3, 4]
y_values = [5, 4, 6, 2]

plt.scatter(x_values, y_values, label="Scatter",color="red")

other_x_values = [1, 2, 3, 4]
other_y_values = [4, 2, 3, 9]

plt.plot(other_x_values, other_y_values, label="Plot", color="green")

plt.title("sample plot")

plt.xlabel("x values")
plt.ylabel("y values")

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data from your script
x_values = [1, 2, 3, 4]
y_1 = [5, 4, 6, 2]
y_2 = [4, 2, 3, 9]

# Set the positions of the bars
x = np.arange(len(x_values)) 
print(x)
width = 0.35  # width of each bar

# Create the bars
plt.bar(x - width/2, y_1, width, label='First Set', color='skyblue')
plt.bar(x + width/2, y_2, width, label='Second Set', color='navy')

# Add labels and formatting
plt.xlabel("x values")
plt.ylabel("y values")
plt.title("Sample Bar Plot")
plt.xticks(x, x_values) # Ensure the x-axis shows 1, 2, 3, 4

plt.legend(loc="upper left")
plt.show()
# plt.savefig('bar_graph.png')
```
