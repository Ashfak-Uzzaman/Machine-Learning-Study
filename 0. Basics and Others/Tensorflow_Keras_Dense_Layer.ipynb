### Simple nural Network with 1-hidden layer using **'Keras'**

````python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from keras.layers import Dense, Input
from keras import Sequential
from keras.losses import MeanSquaredError, BinaryCrossentropy
from keras.activations import sigmoid
````

````python
x_train = np.array([[-0.79173642, 0.63002953]],dtype=np.float32)  # numpy array         #(size in 1000 square feet)
   
````

### A Dense layer
A dense layer is essentially a set of weights and biases applied to the input wrapped in an activation function.

We have a single input x and we would apply a weight w and bias b then wrap that in an activation function a giving us..

```a(wx + b)```

depending on how many units / nodes we want our dense layer to have we can just repeat this with different weights & biases giving us

```
[a(w₀x + b₀),
 a(w₁x + b₁), 
 ... 
 a(wₙx + bₙ)]
```

````python
# 1st layer
layer_1 = Dense(units=25, activation = 'sigmoid')
# 2nd Layer
layer_2=Dense(units=15,activation='sigmoid')
# 3rd Layer
layer_3=Dense(units=1,activation='sigmoid')

model=keras.Sequential(
    [layer_1,
     layer_2,
     layer_3,
    ]
)

y=model(x_train)

print(y)

final_output=1 if y>=.5 else 0
print(f'Final output: {final_output}')

````
