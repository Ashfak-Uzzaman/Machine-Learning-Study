```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import precision_recall_curve, f1_score
import matplotlib.pyplot as plt
import numpy as np

# Create a highly imbalanced dataset
X, y = make_classification(n_samples=5000, n_features=20, weights=[0.95, 0.05], flip_y=0, random_state=42)
# Split into train and test
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, random_state=42)

# Train a classifiers
clf = RandomForestClassifier(random_state=42)
clf.fit(X_train, y_train)

# Predict probabilities for positive class
y_probs = clf.predict_proba(X_test)[:, 1]


precisions, recalls, thresholds = precision_recall_curve(y_test, y_probs)

# Calculate F1 for each threshold
f1_scores = 2 * (precisions * recalls) / (precisions + recalls + 1e-8)  # add small epsilon to avoid div by zero



plt.figure(figsize=(8, 6))
plt.plot(thresholds, precisions[:-1], label='Precision')
plt.plot(thresholds, recalls[:-1], label='Recall')
plt.plot(thresholds, f1_scores[:-1], label='F1 Score')
plt.xlabel('Threshold')
plt.ylabel('Score')
plt.title('Precision, Recall and F1 vs Threshold')
plt.legend()
plt.grid(True)
plt.show()
```

### Slicing
Slicing creates an array of indices using a set of three values (start:stop:step)

```python
#vector slicing operations
a = np.arange(10)
print(f"a         = {a}")

#access 5 consecutive elements (start:stop:step)
c = a[2:7:1];     print("a[2:7:1] = ", c)

# access 3 elements separated by two 
c = a[2:7:2];     print("a[2:7:2] = ", c)

# access all elements index 3 and above
c = a[3:];        print("a[3:]    = ", c)

# access all elements below index 3
c = a[:3];        print("a[:3]    = ", c)

# access all elements
c = a[:];         print("a[:]     = ", c)
```

### Dot product
The dot product is a mainstay of Linear Algebra and NumPy. This is an operation used extensively in this course and should be well understood. The dot product is shown below.

```python
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])

c = np.dot(a, b)

print(c)
print(c.dtype)

c = np.dot(b, a)
print(c)
```

### Indexing
Matrices include a second index. The two indexes describe [row, column]. Access can either return an element or a row/column. See below:

```python
#vector indexing operations on matrices
a = np.arange(6).reshape(-1, 2) # -1,2 means it have 2 columns. 
# So 1st row have 2 column, 2nd row have 2 column,... nth row have 2 column
print(a.shape)

#access an element
print(a[2,0]) # 2nd row, 0th column(element)
print(a[2, 0].shape)
print(type(a[2,0]))



#access a row
print(a[2])
print(a[2].shape)
print(type(a[2]))
```

#### **clip()**

What clip() does:

Given an input array (or similar data structure) and a minimum and maximum value:

Any element in the input that is less than the specified minimum will be replaced by the minimum value.
Any element in the input that is greater than the specified maximum will be replaced by the maximum value.
Elements within the specified range remain unchanged.

```python
a=[1,.5,.003,3,.00000000000012,.00000000001,1.3]
mn = .01
mx=1.0

x=np.clip(a, mn, mx)
x
```
