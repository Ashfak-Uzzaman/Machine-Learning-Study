### imports

```python
import numpy as np
import pandas as pd
```

```python
arr = np.array([[10, 11, 12],
                [13, 14, 15],
                [1, 2, 3],
                [4, 5, 6],
                [7, 8, 9]])
```

### Define the column titles as a list

```python

column_titles = ["Column_A", "Column_B", "Column_C"]
```

### Convert the NumPy array to a Pandas DataFrame

```python

df = pd.DataFrame(arr, columns=column_titles)
```

### Save the DataFrame to a CSV file
'index=False' prevents writing the DataFrame index as a column in the CSV

```python

df.to_csv('output_pandas.csv', index=False)

print("NumPy array converted to DataFrame and saved to output_pandas.csv")
```

### Display CSV File and Dataframe in Terminal

##### Display Dataframe

```python
from tabulate import tabulate
df = pd.DataFrame(arr, columns=column_titles)
print(tabulate(df, headers='keys', tablefmt='grid'))
```

##### Display CSV

```python
import csv
from tabulate import tabulate

def display_csv_tabulate(filepath):
    with open(filepath, 'r', newline='') as csvfile:
        reader = csv.reader(csvfile)
        header = next(reader)
        rows = list(reader)
   
        print(tabulate(rows, headers=header, tablefmt="grid")) 
```

```python
display_csv_tabulate('output_pandas.csv')
```
