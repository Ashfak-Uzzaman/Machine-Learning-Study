##### 9.8.1 List of Tuple Sorting with 'comp function' : Array of Pair Sorting

````python
def comp(a, b):
    if a[0] % 2 == 0:
        return b[0] - a[0]  # Descending order by first element if even
    else:
        return b[1] - a[1]  # Descending order by second element otherwise

# Custom sort key with functools.cmp_to_key
from functools import cmp_to_key

a = [(1, 2), (3, 4), (5, 6)]

# Sort using custom comparator
a.sort(key=cmp_to_key(comp))

# Optional: Print sorted list
for pair in a:
    print(pair)
````

##### **16. Heaps : (multiset in c++)**
**'min-heap' (by default )**

````python
# Heaps
import heapq
# By default min-heap

# under the hood are arrays
minHeap = []
heapq.heappush(minHeap, 3)
heapq.heappush(minHeap, 2)
heapq.heappush(minHeap, 4)

# Min is always at index 0
print(minHeap[0])

while len(minHeap):
    print(heapq.heappop(minHeap))
````

**'max-heap': (-1) * min-heap** 

````python
# Heaps
import heapq

minHeap = []
heapq.heappush(minHeap, -1*3)
heapq.heappush(minHeap, -1*2)
heapq.heappush(minHeap, -1*4)

# Min is always at index 0
print(-1*minHeap[0])

while len(minHeap):
    print(-1*heapq.heappop(minHeap))
````

````python
arr = [2,1,8,4,5]
heapq.heapify(arr)

while arr:
    print(heapq.heappop(arr))
````

##### **18. Class**

````python
# Class
class MyClass:
    # Constructor
    def __init__(self, nums):
        # Create member variables
        self.nums = nums
        self.size = len(nums)

    # self key word required as param
    def getLength(self):
        return self.size

    def getDoubleLength(self):
        return 2 * self.getLength()
````

````python
# Create an object (instance) of MyClass
# We pass a list [10, 20, 30, 40] to the constructor (which becomes the 'nums' argument)
my_object = MyClass([10, 20, 30, 40])

# Access member variables
print(my_object.nums)
print(my_object.size)

# Call methods on the object
length = my_object.getLength()
double_length = my_object.getDoubleLength()

print(length)
print(double_length)

# We can also create another object with different data
another_object = MyClass([5, 10])
print(another_object.nums)
print(another_object.getLength())
````
